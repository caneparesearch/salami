```python
import os
from pymatgen.core.structure import Structure
from salami.generator import SlabReconstructor

file_dir=os.path.join(".")


conv_structure=Structure.from_file(os.path.join(file_dir,"SrTiO3.cif"),primitive=False)    



slabgen=SlabReconstructor(
    initial_structure=conv_structure,
    oxidation_states={
        "Sr":2,
        "Ti":4,
        "O":-2
        },
    
)
slabgen.set_valid_criteria(
        bonds_and_coordination=[
        (
            {
                ("Sr2+","O2-"):(2.8,1,12),
                ("Ti4+",'O2-'):(2.1,6,6),
            },
        ),
        (
            {
                ("O2-","Sr2+"):(2.8,1,12)
            },
            {
                ("O2-","Ti4+"):(2.1,1,6)
            }
        ),

        ],
        criteria={"pass_coordination_number_test":True,
                "is_polar":False,
                "is_symmetric":True,
                "charge_neutral":True
                },
        )
    
```

```python

slabgen.generate_initial_slabs(
    miller_index=[0,0,1],
    min_slab_size=25,
    min_vacuum_size=15,
    in_unit_planes=False,
    )
    
```

```python

slabgen.generate_symmetrified_slabs(
        from_slab_pool=["initial_orthogonal_slabs"],
        partial_explore_removable_depth=1.0,
        filter_type="LowestOne",
        filter_kwargs={},
        )

print('valid slabs:',len(slabgen._obtain_from_slabpool(
    from_slab_pool=['valid_slabs'],
    deepcopy=True,
    )),slabgen._obtain_from_slabpool(
    from_slab_pool=['valid_slabs'],
    deepcopy=True,
    ))
```
