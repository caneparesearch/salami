```python
from pymatgen.core.composition import Composition
from pymatgen.core.periodic_table import Species
from pymatgen.core.structure import Structure

structure=Structure.from_file("files/cubic-Na3PS4/prim.cif")
print(structure)
```

```python

print(structure[0].species,type(structure[0].species))
```

```python
structure.add_oxidation_state_by_element({
    "Na":1,
    "P":5,
    "S":-2,
})
print(structure[0].species,type(structure[0].species))
```

```python

print("Na+" in structure[0].species,"Na" in structure[0].species)
print("Na+" in structure[0],"Na" in structure[0])
print("Na+" in structure[0].species_string,"Na" in structure[0].species_string)
#print('Na+' in structure, 'Na' in structure)
print("Na+" in structure.composition,"Na" in structure.composition)
print(Species("Na")==structure[0].species,Species("Na",oxidation_state=1)==structure[0].species)
print(Composition("Na")==structure[0].species,Composition({Species('Na',1):1},)==structure[0].species)
```
