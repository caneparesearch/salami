stoichiometric formula: charged ion and ion without charge lead descrepancy 

```python
from pymatgen.core.structure import Structure
s=Structure.from_file("files/linh/linh_symmetrified.cif")
print(s.composition.get_reduced_composition_and_factor()[0])
b=Structure.from_file("files/linh/linh_symmetrified.cif")
b.add_oxidation_state_by_element({"Li": 1, "N": -3, "H": 1})
print(b.composition.get_reduced_composition_and_factor()[0])

sc=s.composition.get_reduced_composition_and_factor()[0]
bc=b.composition.get_reduced_composition_and_factor()[0]
print((sc,bc))
print(sc==bc)
print((bc==sc))
print(sc.almost_equals(bc))
```

```python
neutral_comp = sc.element_composition

print(neutral_comp)
neutral2 = bc.element_composition
print(neutral2)
print(neutral_comp.almost_equals(neutral2))
print(neutral_comp == neutral2)
print(neutral2 == neutral_comp)
```
